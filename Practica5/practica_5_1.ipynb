```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# Crear lienzo
img = np.full((600, 600, 3), (180,220,255), dtype=np.uint8)

# -------------------------------------------------
# COLORES (BGR)
# -------------------------------------------------

body = (40,70,150)
neck = (30,50,120)
face = (120,170,220)
eye = (255,255,255)
pupil = (0,0,0)
ear = (40,60,130)
grass = (80,180,80)

# -------------------------------------------------
# PASTO
# -------------------------------------------------

cv2.rectangle(img,(0,420),(600,600),grass,-1)

# -------------------------------------------------
# CUERPO
# -------------------------------------------------

cv2.ellipse(img,(300,430),(120,80),0,0,360,body,-1)

# -------------------------------------------------
# CUELLO
# -------------------------------------------------

cv2.rectangle(img,(280,220),(320,420),neck,-1)

# -------------------------------------------------
# CABEZA
# -------------------------------------------------

cv2.ellipse(img,(300,200),(70,55),0,0,360,face,-1)

# -------------------------------------------------
# OREJAS
# -------------------------------------------------

pts1 = np.array([[260,140],[280,200],[300,150]])
cv2.fillPoly(img,[pts1],ear)

pts2 = np.array([[340,140],[320,200],[300,150]])
cv2.fillPoly(img,[pts2],ear)

# -------------------------------------------------
# OJOS
# -------------------------------------------------

cv2.circle(img,(280,200),12,eye,-1)
cv2.circle(img,(320,200),12,eye,-1)

cv2.circle(img,(280,200),6,pupil,-1)
cv2.circle(img,(320,200),6,pupil,-1)

# -------------------------------------------------
# NARIZ
# -------------------------------------------------

cv2.ellipse(img,(300,230),(20,10),0,0,360,(60,80,120),-1)

# -------------------------------------------------
# PATAS
# -------------------------------------------------

cv2.rectangle(img,(250,420),(270,520),body,-1)
cv2.rectangle(img,(330,420),(350,520),body,-1)

# -------------------------------------------------
# COLA
# -------------------------------------------------

cv2.circle(img,(420,420),20,body,-1)

# -------------------------------------------------
# MOSTRAR IMAGEN
# -------------------------------------------------

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6,6))
plt.imshow(img)
plt.axis("off")
plt.title("Llama estilo Kuzco")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


height, width = 450, 720
img = np.zeros((height, width, 3), dtype=np.uint8)

# Fondo celeste (igual que tu código)
img[:, :] = [100, 180, 255]
img[320:, :] = [40, 160, 60]  # césped verde abajo


def dibujar_pokemon_rojo(x, y):
    """Pokémon tipo bola roja con ojos grandes."""
    color_body = [230, 80, 80]       # rojo suave
    color_eye_white = [255, 255, 255]
    color_eye_black = [20, 20, 20]
    color_mouth = [200, 80, 80]

    # Cuerpo grande (aprox. círculo con cuadrados)
    img[y+ 0:y+40, x+20:x+40] = color_body
    img[y+ 0:y+40, x+40:x+60] = color_body
    img[y+ 0:y+40, x+60:x+80] = color_body

    img[y+40:y+80, x+10:x+30] = color_body
    img[y+40:y+80, x+30:x+50] = color_body
    img[y+40:y+80, x+50:x+70] = color_body
    img[y+40:y+80, x+70:x+90] = color_body

    img[y+80:y+120, x+20:x+40] = color_body
    img[y+80:y+120, x+40:x+60] = color_body
    img[y+80:y+120, x+60:x+80] = color_body

    # Ojo izquierdo
    img[y+50:y+70, x+35:x+55] = color_eye_white
    img[y+55:y+65, x+40:x+50] = color_eye_black

    # Ojo derecho
    img[y+50:y+70, x+55:x+75] = color_eye_white
    img[y+55:y+65, x+60:x+70] = color_eye_black

    # Boca
    img[y+90:y+95, x+45:x+65] = color_mouth


def dibujar_pokemon_azul(x, y):
    """Pokémon tipo bola azul con antenas."""
    color_body = [80, 140, 240]      # azul suave
    color_antenna = [90, 170, 250]
    color_eye_white = [255, 255, 255]
    color_eye_black = [20, 20, 20]

    # Cuerpo grande
    img[y+ 0:y+40, x+25:x+45] = color_body
    img[y+ 0:y+40, x+45:x+65] = color_body
    img[y+ 0:y+40, x+65:x+85] = color_body

    img[y+40:y+80, x+15:x+35] = color_body
    img[y+40:y+80, x+35:x+55] = color_body
    img[y+40:y+80, x+55:x+75] = color_body
    img[y+40:y+80, x+75:x+95] = color_body

    img[y+80:y+120, x+25:x+45] = color_body
    img[y+80:y+120, x+45:x+65] = color_body
    img[y+80:y+120, x+65:x+85] = color_body

    # Antena izquierda
    img[y-10:y+10, x+38:x+42] = color_antenna

    # Antena derecha
    img[y-10:y+10, x+68:x+72] = color_antenna

    # Ojos
    img[y+50:y+70, x+40:x+50] = color_eye_white
    img[y+55:y+65, x+42:x+48] = color_eye_black

    img[y+50:y+70, x+60:x+70] = color_eye_white
    img[y+55:y+65, x+62:x+68] = color_eye_black


def dibujar_pokemon_amarillo(x, y):
    """Pokémon tipo “bola amarilla” con orejas grandes."""
    color_body = [255, 220, 100]     # amarillo suave
    color_ear = [255, 200, 80]
    color_eye_white = [255, 255, 255]
    color_eye_black = [20, 20, 20]

    # Oreja izquierda
    img[y+ 0:y+20, x+20:x+30] = color_ear
    img[y+20:y+40, x+25:x+35] = color_ear

    # Oreja derecha
    img[y+ 0:y+20, x+70:x+80] = color_ear
    img[y+20:y+40, x+65:x+75] = color_ear

    # Cuerpo grande
    img[y+40:y+80, x+25:x+45] = color_body
    img[y+40:y+80, x+45:x+65] = color_body
    img[y+40:y+80, x+65:x+85] = color_body

    img[y+80:y+120, x+30:x+50] = color_body
    img[y+80:y+120, x+50:x+70] = color_body
    img[y+80:y+120, x+70:x+90] = color_body

    # Ojos
    img[y+55:y+75, x+40:x+50] = color_eye_white
    img[y+60:y+70, x+42:x+48] = color_eye_black

    img[y+55:y+75, x+60:x+70] = color_eye_white
    img[y+60:y+70, x+62:x+68] = color_eye_black


def dibujar_pokemon_verde(x, y):
    """Pokémon tipo “planta” con formas rectangulares."""
    color_body = [100, 190, 100]     # verde suave
    color_leaf = [80, 160, 80]
    color_eye_white = [255, 255, 255]
    color_eye_black = [20, 20, 20]

    # Hojas superiores
    img[y+ 0:y+20, x+30:x+40] = color_leaf
    img[y+ 0:y+20, x+40:x+50] = color_leaf
    img[y+ 0:y+20, x+50:x+60] = color_leaf

    # Cuerpo grande
    img[y+20:y+60, x+30:x+50] = color_body
    img[y+20:y+60, x+50:x+70] = color_body

    img[y+60:y+100, x+25:x+45] = color_body
    img[y+60:y+100, x+45:x+65] = color_body
    img[y+60:y+100, x+65:x+85] = color_body

    # Ojos
    img[y+35:y+55, x+40:x+50] = color_eye_white
    img[y+40:y+50, x+42:x+48] = color_eye_black

    img[y+35:y+55, x+55:x+65] = color_eye_white
    img[y+40:y+50, x+57:x+63] = color_eye_black


# Dibujar varios Pokémon en posiciones distintas
dibujar_pokemon_rojo( 80,  90)   # izquierda alta
dibujar_pokemon_azul(240, 110)   # centro-alto
dibujar_pokemon_amarillo(410, 120) # derecha alta
dibujar_pokemon_verde(580, 150)  # derecha un poco más abajo


# Mostrar la imagen
plt.figure(figsize=(14, 8))
plt.imshow(img)
plt.axis('off')
plt.title("Imagen con varios Pokémon solo con cuadrados")
plt.tight_layout()
plt.show()

# Guardar la imagen
plt.imsave('pokemones_cuadrados.png', img)
print("Imagen guardada como 'pokemones_cuadrados.png'")
```
