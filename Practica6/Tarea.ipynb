Tarea de vertical horizontal diagonal derecha y diagonal izquierda

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

#Convolución Media horizontal
def convolucion_media_h(imagen,tkernel):
    kernel = np.ones((1,tkernel),np.float32)/tkernel
    imagen_c = cv2.filter2D(imagen,-1,kernel)
    return imagen_c

#Convolución Media vertical
def convolucion_media_v(imagen,tkernel):
    kernel = np.ones((tkernel,1),np.float32)/tkernel
    imagen_c = cv2.filter2D(imagen,-1,kernel)
    return imagen_c

#Convoluvion Media diagonal
def convolucion_media_d(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)

    for i in range(tkernel):
        kernel[i][i] = 1

    kernel = kernel / tkernel

    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

#Convolucion Media diagonal inversa
def convolucion_media_di(imagen,tkernel):
    kernel = np.zeros((tkernel,tkernel),np.float32)
    for i in range(tkernel):
        kernel[i][tkernel-1-i] = 1
    kernel = kernel/tkernel
    imagen_c = cv2.filter2D(imagen,-1,kernel)
    return imagen_c




img = cv2.imread('BMW_635.jpg')

img_fin1 = convolucion_media_h(img,101)
img_fin2 = convolucion_media_v(img,101)
img_fin3 = convolucion_media_d(img,101)
img_fin4 = convolucion_media_di(img,101)



plt.figure(figsize=(15, 12))
plt.subplot(3,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')

plt.subplot(3,2,2)
plt.imshow(cv2.cvtColor(img_fin1, cv2.COLOR_BGR2RGB))
plt.title('Convolución Media Horizontal')

plt.subplot(3,2,3)
plt.imshow(cv2.cvtColor(img_fin2, cv2.COLOR_BGR2RGB))
plt.title('Convolución Media Vertical')

plt.subplot(3,2,4)
plt.imshow(cv2.cvtColor(img_fin3, cv2.COLOR_BGR2RGB))
plt.title('Convolución Media Diagonal')

plt.subplot(3,2,5)
plt.imshow(cv2.cvtColor(img_fin4, cv2.COLOR_BGR2RGB))
plt.title('Convolución Media Diagonal Inversa')

plt.show()



```
