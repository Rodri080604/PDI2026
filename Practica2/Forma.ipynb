```python
import cv2
import matplotlib.pyplot as plt

# Cargar foto 1 en BGR (formato normal de OpenCV)
img1 = cv2.imread('foto1.jpeg')          

# Convertir a RGB para que matplotlib la muestre bien
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

# Mostrar foto 1 en color
plt.figure(figsize=(8,6))
plt.title('Foto 1 - Color original')
plt.imshow(img1_rgb)
plt.show()

# Convertir a escala de grises
img1_gray = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

# Mostrar en gris
plt.figure(figsize=(8,6))
plt.title('Foto 1 - Escala de grises')
plt.imshow(img1_gray, cmap='gray')
plt.show()

# Guardar ambas versiones
cv2.imwrite('FOTO1_ORIGINAL.jpg', img1)
cv2.imwrite('FOTO1_GRISjpg', img1_gray)
print("Foto 1 guardada (original y gris)")
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('foto1.jpeg')           # ← mismo nombre que usaste arriba
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Separar canales
b = img[:,:,0]
g = img[:,:,1]
r = img[:,:,2]

# Color falso (ejemplo JET)
falso_color = cv2.applyColorMap(gray, cv2.COLORMAP_JET)
falso_rgb = cv2.cvtColor(falso_color, cv2.COLOR_BGR2RGB)

# Mostrar todo junto
plt.figure(figsize=(15, 8))

plt.subplot(2,3,1)
plt.title('Color original')
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(2,3,2)
plt.title('Gris')
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(2,3,3)
plt.title('Canal Azul')
plt.imshow(b, cmap='gray')
plt.axis('off')

plt.subplot(2,3,4)
plt.title('Canal Verde')
plt.imshow(g, cmap='gray')
plt.axis('off')

plt.subplot(2,3,5)
plt.title('Canal Rojo')
plt.imshow(r, cmap='gray')
plt.axis('off')

plt.subplot(2,3,6)
plt.title('Color falso (JET)')
plt.imshow(falso_rgb)
plt.axis('off')

plt.suptitle('Foto 1 - Análisis completo')
plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('foto2.jpeg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

b = img[:,:,0]
g = img[:,:,1]
r = img[:,:,2]

falso_color = cv2.applyColorMap(gray, cv2.COLORMAP_VIRIDIS)
falso_rgb = cv2.cvtColor(falso_color, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 8))

plt.subplot(2,3,1); plt.imshow(img_rgb); plt.title('Color original'); plt.axis('off')
plt.subplot(2,3,2); plt.imshow(gray, cmap='gray'); plt.title('Gris'); plt.axis('off')
plt.subplot(2,3,3); plt.imshow(b, cmap='gray'); plt.title('Azul'); plt.axis('off')
plt.subplot(2,3,4); plt.imshow(g, cmap='gray'); plt.title('Verde'); plt.axis('off')
plt.subplot(2,3,5); plt.imshow(r, cmap='gray'); plt.title('Rojo'); plt.axis('off')
plt.subplot(2,3,6); plt.imshow(falso_rgb); plt.title('Color falso (VIRIDIS)'); plt.axis('off')

plt.suptitle('Foto 2 - Análisis completo')
plt.tight_layout()
plt.show()
```
