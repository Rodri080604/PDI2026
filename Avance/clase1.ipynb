```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize']=[10,5]
img=cv2.imread('BMW_635.jpg',cv2.IMREAD_COLOR)
imgRGB=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

r=img[:,:,2]
g=img[:,:,1]
b=img[:,:,0]

#crear imagen por canal
r_img=np.zeros_like(imgRGB)
r_img[:,:,0]=r

g_img=np.zeros_like(imgRGB)
g_img[:,:,1]=g

b_img=np.zeros_like(imgRGB)
b_img[:,:,2]=b

plt.figure(figsize=(10,5))


plt.subplot(141)
plt.title('CANAL ORIGINAL')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('CANAL ROJO')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('CANAL VERDE')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title('CANAL AZUL')
plt.imshow(b_img)
plt.axis('off')

```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [16, 9]  


img = cv2.imread('BMW_635.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


r = imgRGB[:,:,0]   
g = imgRGB[:,:,1]   
b = imgRGB[:,:,2]   


r_img = np.zeros_like(imgRGB); r_img[:,:,0] = r
g_img = np.zeros_like(imgRGB); g_img[:,:,1] = g
b_img = np.zeros_like(imgRGB); b_img[:,:,2] = b


y_img = np.zeros_like(imgRGB); y_img[:,:,0] = r; y_img[:,:,1] = g          
c_img = np.zeros_like(imgRGB); c_img[:,:,1] = g; c_img[:,:,2] = b          
m_img = np.zeros_like(imgRGB); m_img[:,:,0] = r; m_img[:,:,2] = b          
w_img = np.zeros_like(imgRGB); w_img[:,:,0] = r; w_img[:,:,1] = g; w_img[:,:,2] = b  
k_img = np.zeros_like(imgRGB)  


colores_fila2 = [r_img, g_img, b_img, y_img, c_img]
nombres_fila2 = ['Rojo', 'Verde', 'Azul', 'Amarillo', 'Cian']

colores_fila3 = [m_img, w_img, k_img, y_img, m_img]   
nombres_fila3 = ['Magenta', 'Blanco', 'Negro', 'Amarillo', 'Magenta']


plt.figure(figsize=(16, 10))

# Fila 1 - Imagen original
plt.subplot(3, 5, 1)
plt.title('Original')
plt.imshow(imgRGB)
plt.axis('off')

# Espacios vacíos en la primera fila (para que quede centrado o alineado)
for i in range(2, 6):
    plt.subplot(3, 5, i)
    plt.axis('off')

# Fila 2 - 5 colores
for i, (img_color, nombre) in enumerate(zip(colores_fila2, nombres_fila2), start=1):
    plt.subplot(3, 5, 5 + i)
    plt.title(nombre)
    plt.imshow(img_color)
    plt.axis('off')

# Fila 3 - otros 5 colores
for i, (img_color, nombre) in enumerate(zip(colores_fila3, nombres_fila3), start=1):
    plt.subplot(3, 5, 10 + i)
    plt.title(nombre)
    plt.imshow(img_color)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = [16, 9]

# Leer imagen
img = cv2.imread('BMW_635.jpg', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Canales
r = imgRGB[:,:,0]  # rojo
g = imgRGB[:,:,1]  # verde
b = imgRGB[:,:,2]  

# Imágenes de un solo canal
r_img = np.zeros_like(imgRGB); r_img[:,:,0] = r
g_img = np.zeros_like(imgRGB); g_img[:,:,1] = g
b_img = np.zeros_like(imgRGB); b_img[:,:,2] = b


y_img  = np.zeros_like(imgRGB); y_img[:,:,0] = r; y_img[:,:,1]  = g           
c_img  = np.zeros_like(imgRGB); c_img[:,:,1] = g; c_img[:,:,2]  = b           
m_img  = np.zeros_like(imgRGB); m_img[:,:,0] = r; m_img[:,:,2]  = b           


orange_img = np.zeros_like(imgRGB); orange_img[:,:,0] = r; orange_img[:,:,1] = g * 0.65    
purple_img = np.zeros_like(imgRGB); purple_img[:,:,0] = r * 0.75; purple_img[:,:,2] = b * 0.9   
pink_img   = np.zeros_like(imgRGB); pink_img[:,:,0]   = r;   pink_img[:,:,1]   = g * 0.4; pink_img[:,:,2] = b * 0.7   
brown_img  = np.zeros_like(imgRGB); brown_img[:,:,0]  = r * 0.6; brown_img[:,:,1]  = g * 0.4; brown_img[:,:,2] = b * 0.2   
gray_img   = np.zeros_like(imgRGB); gray_img[:,:,0]   = r * 0.5; gray_img[:,:,1]   = g * 0.5; gray_img[:,:,2] = b * 0.5   


plt.figure(figsize=(16, 10))

# Fila 1 
plt.subplot(3, 5, 1)
plt.title('Original')
plt.imshow(imgRGB)
plt.axis('off')

# Espacios vacíos en fila 1
for i in range(2, 6):
    plt.subplot(3, 5, i)
    plt.axis('off')

# Fila 2 
colores_fila2 = [r_img, g_img, b_img, y_img, c_img]
nombres_fila2 = ['Rojo', 'Verde', 'Azul', 'Amarillo', 'Cian']

for i, (img_c, nom) in enumerate(zip(colores_fila2, nombres_fila2), start=1):
    plt.subplot(3, 5, 5 + i)
    plt.title(nom)
    plt.imshow(img_c)
    plt.axis('off')

# Fila 3 
colores_fila3 = [orange_img, purple_img, pink_img, brown_img, gray_img]
nombres_fila3 = ['Naranja', 'Morado', 'Rosa', 'Marrón', 'Gris']

for i, (img_c, nom) in enumerate(zip(colores_fila3, nombres_fila3), start=1):
    plt.subplot(3, 5, 10 + i)
    plt.title(nom)
    plt.imshow(img_c)
    plt.axis('off')

plt.tight_layout()
plt.show()
```
