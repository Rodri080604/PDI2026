SOLARIZACION

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def solarizar(imagen, umbral):
    # APLICAMOS SOLARIZACION A LA IMAGEN
    imagen_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return imagen_sol
    
imagen = cv2.imread('BMW_635.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)  # Convertir a escala de grises


umbral = 128  # Definir el umbral para la solarización
img_fin=solarizar(img_gray, umbral)  # Aplicar la función de solarización


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 2, 2)
plt.title('Imagen Solarizada en escala de grises')
plt.imshow(img_fin, cmap='gray')  # Mostrar la imagen solarizada en escala de grises
plt.show()




```

POSTERIZACION

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def posterizar(imagen, niveles):
    # APLICAMOS POSTERIZACION A LA IMAGEN
    factor_escala = 256 // niveles
    imagen_post = np.round(imagen / factor_escala) * factor_escala
    return imagen_post.astype(np.uint8) # Convertir a uint8 para mantener el formato de imagen


imagen = cv2.imread('BMW_635.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)  # Convertir a escala de grises

nvles = 3  # Definir el número de niveles para la posterización
img_fin=posterizar(img_gray, nvles)  # Aplicar la función de posterización 



plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 2, 2)
plt.title('Imagen Posterizada en escala de grises')
plt.imshow(img_fin, cmap='gray')  # Mostrar la imagen posterizada en escala de grises
plt.show()

```

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt


def posterizar(imagen, niveles):
    # APLICAMOS POSTERIZACION A LA IMAGEN
    factor_escala = 256 // niveles
    imagen_post = np.round(imagen / factor_escala) * factor_escala
    return imagen_post.astype(np.uint8) # Convertir a uint8 para mantener el formato de imagen


def solarizar(imagen, umbral):
    # APLICAMOS SOLARIZACION A LA IMAGEN
    imagen_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return imagen_sol


imagen = cv2.imread('BMW_635.jpg') 
umbral = 50  # Definir el umbral para la solarización
img_fin=solarizar(imagen, umbral)  # Aplicar la función de solarización

nvles = 3  # Definir el número de niveles para la posterización
img2_fin=posterizar(imagen, nvles)  # Aplicar la función de posterización


plt.figure(figsize=(12, 6))
plt.subplot(1, 3, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 3, 2)
plt.title('Imagen Solarizada')
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))  # Mostrar la imagen solarizada

plt.subplot(1, 3, 3)
plt.title('Imagen Posterizada')
plt.imshow(cv2.cvtColor(img2_fin, cv2.COLOR_BGR2RGB))  # Mostrar la imagen posterizada
plt.show()
```

FALSO COLOR

```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt

def falso_color(img_gray):
    escal_img=cv2.normalize(img_gray,None,0,255,cv2.NORM_MINMAX)
    falso_img=cv2.applyColorMap(escal_img,cv2.COLORMAP_JET)
    return falso_img

imagen = cv2.imread('BMW_635.jpg')  # Cargar la imagen en color 
img_gray=cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY) 
img_fin=falso_color(img_gray)   

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 2)
plt.title('Imagen Falso Color')
plt.imshow(cv2.cvtColor(img_fin, cv2.COLOR_BGR2RGB))  # Mostrar la imagen original

plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(img_gray, cmap='gray')  # Mostrar la imagen solarizada en escala de grises
plt.show()
```
