CONVOLUCIONES CON MATRIZ KERNEL

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('Fond.jpg',0)

#MASCARA DE CONVOLUCION
kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
img_fin = cv2.filter2D(img,-1,kernel)

plt.subplot(1,2,1)
plt.imshow(img,cmap = 'gray')

plt.subplot(1,2,2)
plt.imshow(img_fin,cmap = 'gray')
plt.show()


```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('BMW_635.jpg',0)

kernel = np.ones((5,5),np.float32)/25
suavizada = cv2.filter2D(img,-1,kernel)

kernel2 = np.ones((7,7),np.float32)/49
suavizada2 = cv2.filter2D(img,-1,kernel2)

plt.figure(figsize=(15, 8))
plt.subplot(1,3,1)
plt.imshow(img,cmap = 'gray')
plt.title('Imagen Original')

plt.subplot(1,3,2)
plt.imshow(suavizada,cmap = 'gray')
plt.title('Suavizada con kernel 5x5')

plt.subplot(1,3,3)
plt.imshow(suavizada2,cmap = 'gray')
plt.title('Suavizada con kernel 7x7')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('BMW_635.jpg')

kernel = np.ones((10,10),np.float32)/100
suavizada = cv2.filter2D(img,-1,kernel)

kernel2 = np.ones((12,12),np.float32)/144
suavizada2 = cv2.filter2D(img,-1,kernel2)

plt.figure(figsize=(15, 8))
plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')

plt.subplot(1,3,2)
plt.imshow(cv2.cvtColor(suavizada, cv2.COLOR_BGR2RGB))
plt.title('Suavizada')

plt.subplot(1,3,3)
plt.imshow(cv2.cvtColor(suavizada2, cv2.COLOR_BGR2RGB))
plt.title('Suavizada')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('BMW_635.jpg',0)

kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
enfoque = cv2.filter2D(img,-1,kernel)

kernel2 = np.array([[1,1,1],[1,1,1],[1,1,1]])
desenfoque = cv2.filter2D(img,-1,kernel2)

kernel3 = np.array([[0,0,0],[-1,1,0],[0,0,0]])
Realce_de_bordes = cv2.filter2D(img,-1,kernel3)

kernel4 = np.array([[-2,-1,0],[-1,1,1],[0,1,2]])
Repujado = cv2.filter2D(img,-1,kernel4)

kernel5 = np.array([[0,1,0],[1,-4,1],[0,1,0]])
deteccion_de_bordes = cv2.filter2D(img,-1,kernel5)

kernel6 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
filtro_de_tipo_sobel = cv2.filter2D(img,-1,kernel6)

kernel7 = np.array([[1,-2,1],[-2,5,-2],[1,-2,1]])
filtro_de_tipo_sharpen = cv2.filter2D(img,-1,kernel7)

kernel8 = np.array([[1,1,1],[1,-2,1],[-1,-1,-1]])
filtro_norte = cv2.filter2D(img,-1,kernel8)

kernel9 = np.array([[-1,1,1],[-1,-2,1],[-1,1,1]])
filtro_este = cv2.filter2D(img,-1,kernel9)

kernel10 = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,11,3],[2,7,11,7,1],[1,2,3,2,1]])
filtro_de_tipo_gauss = cv2.filter2D(img,-1,kernel10)

plt.figure(figsize=(15, 8))
plt.subplot(3,4,1)
plt.imshow(img,cmap = 'gray')
plt.title('Imagen Original')

plt.subplot(3,4,2)
plt.imshow(enfoque,cmap = 'gray')
plt.title('Enfoque')

plt.subplot(3,4,3)
plt.imshow(desenfoque,cmap = 'gray')
plt.title('Desenfoque')

plt.subplot(3,4,4)
plt.imshow(Realce_de_bordes,cmap = 'gray')
plt.title('Realce de Bordes')

plt.subplot(3,4,5)
plt.imshow(Repujado,cmap = 'gray')
plt.title('Repujado')

plt.subplot(3,4,6)
plt.imshow(deteccion_de_bordes,cmap = 'gray')
plt.title('Detección de Bordes')

plt.subplot(3,4,7)
plt.imshow(filtro_de_tipo_sobel,cmap = 'gray')
plt.title('Filtro de Tipo Sobel')

plt.subplot(3,4,8)
plt.imshow(filtro_de_tipo_sharpen,cmap = 'gray')
plt.title('Filtro de Tipo Sharpen')

plt.subplot(3,4,9)
plt.imshow(filtro_norte,cmap = 'gray')
plt.title('Filtro Norte')

plt.subplot(3,4,10)
plt.imshow(filtro_este,cmap = 'gray')
plt.title('Filtro Este')

plt.subplot(3,4,11)
plt.imshow(filtro_de_tipo_gauss,cmap = 'gray')
plt.title('Filtro de Tipo Gauss')


plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('BMW_635.jpg')

kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
enfoque = cv2.filter2D(img,-1,kernel)

kernel2 = np.array([[1,1,1],[1,1,1],[1,1,1]])
desenfoque = cv2.filter2D(img,-1,kernel2)

kernel3 = np.array([[0,0,0],[-1,1,0],[0,0,0]])
Realce_de_bordes = cv2.filter2D(img,-1,kernel3)

kernel4 = np.array([[-2,-1,0],[-1,1,1],[0,1,2]])
Repujado = cv2.filter2D(img,-1,kernel4)

kernel5 = np.array([[0,1,0],[1,-4,1],[0,1,0]])
deteccion_de_bordes = cv2.filter2D(img,-1,kernel5)

kernel6 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
filtro_de_tipo_sobel = cv2.filter2D(img,-1,kernel6)

kernel7 = np.array([[1,-2,1],[-2,5,-2],[1,-2,1]])
filtro_de_tipo_sharpen = cv2.filter2D(img,-1,kernel7)

kernel8 = np.array([[1,1,1],[1,-2,1],[-1,-1,-1]])
filtro_norte = cv2.filter2D(img,-1,kernel8)

kernel9 = np.array([[-1,1,1],[-1,-2,1],[-1,1,1]])
filtro_este = cv2.filter2D(img,-1,kernel9)

kernel10 = np.array([[1,2,3,1,1],[2,7,11,7,2],[3,11,17,11,3],[2,7,11,7,1],[1,2,3,2,1]])
filtro_de_tipo_gauss = cv2.filter2D(img,-1,kernel10)

plt.figure(figsize=(15, 8))
plt.subplot(3,4,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')

plt.subplot(3,4,2)
plt.imshow(cv2.cvtColor(enfoque, cv2.COLOR_BGR2RGB))
plt.title('Enfoque')

plt.subplot(3,4,3)
plt.imshow(cv2.cvtColor(desenfoque, cv2.COLOR_BGR2RGB))
plt.title('Desenfoque')

plt.subplot(3,4,4)
plt.imshow(cv2.cvtColor(Realce_de_bordes, cv2.COLOR_BGR2RGB))
plt.title('Realce de Bordes')

plt.subplot(3,4,5)
plt.imshow(cv2.cvtColor(Repujado, cv2.COLOR_BGR2RGB))
plt.title('Repujado')

plt.subplot(3,4,6)
plt.imshow(cv2.cvtColor(deteccion_de_bordes, cv2.COLOR_BGR2RGB))
plt.title('Detección de Bordes')

plt.subplot(3,4,7)
plt.imshow(cv2.cvtColor(filtro_de_tipo_sobel, cv2.COLOR_BGR2RGB))
plt.title('Filtro de Tipo Sobel')

plt.subplot(3,4,8)
plt.imshow(cv2.cvtColor(filtro_de_tipo_sharpen, cv2.COLOR_BGR2RGB))
plt.title('Filtro de Tipo Sharpen')

plt.subplot(3,4,9)
plt.imshow(cv2.cvtColor(filtro_norte, cv2.COLOR_BGR2RGB))
plt.title('Filtro Norte')

plt.subplot(3,4,10)
plt.imshow(cv2.cvtColor(filtro_este, cv2.COLOR_BGR2RGB))
plt.title('Filtro Este')

plt.subplot(3,4,11)
plt.imshow(cv2.cvtColor(filtro_de_tipo_gauss, cv2.COLOR_BGR2RGB))
plt.title('Filtro de Tipo Gauss')


plt.show()
```
