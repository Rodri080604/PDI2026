```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = (8, 8) 
size = 200          
half = size // 2    
imagen = np.zeros((size, size, 3), dtype=np.uint8)
for i in range(size):
    for j in range(size):
        if i < half and j < half:
            imagen[i, j] = [0, 125, 0]
        elif i < half and j >= half:
            imagen[i, j] = [0, 0, 255]     
        elif i >= half and j < half:
            imagen[i, j] = [255, 0, 0]     
        else:
            imagen[i, j] = [255, 255, 0] 

plt.figure()
plt.imshow(imagen)
plt.title('4 colores')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = (8, 8)
size = 150
third = size // 3           
imagen = np.zeros((size, size, 3), dtype=np.uint8)

for i in range(size):
    for j in range(size):
        if i < third:
            if j < third:
                imagen[i, j] = [0, 190, 0]
            elif j < 2*third:
                imagen[i, j] = [0, 0, 255]
            else:
                imagen[i, j] = [255, 255, 255]
        elif i < 2*third:
            if j < third:
                imagen[i, j] = [255, 0, 0]
            elif j < 2*third:
                imagen[i, j] = [255, 255, 0]
            else:
                imagen[i, j] = [255, 165, 0]
        else:
            if j < third:
                imagen[i, j] = [255, 255, 255]    
            elif j < 2*third:
                imagen[i, j] = [0, 190, 0]    
            else:
                imagen[i, j] = [0, 0, 255]  

plt.figure()
plt.imshow(imagen)
plt.title('Cubo Rubik')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = (8, 8)
size = 150
third = size // 3

colores_rubik = np.array([
    [0,   190, 0],
    [0,   0,   255],
    [255, 255, 255],
    [255, 0,   0],
    [255, 255, 0],
    [255, 165, 0]
], dtype=np.uint8)

colores_aleatorios = np.random.choice(range(6), size=(3,3))

imagen = np.zeros((size, size, 3), dtype=np.uint8)

for i in range(size):
    for j in range(size):
        fila    = i // third
        columna = j // third
        idx_color = colores_aleatorios[fila, columna]
        imagen[i, j] = colores_rubik[idx_color]

plt.figure()
plt.imshow(imagen)
plt.title('Cubo Rubik – mezcla aleatoria')
plt.show()
```
