```python
import cv2
import matplotlib.pyplot as plt

# Leer imagen
imagen = cv2.imread("Foto1.jpeg")

# Convertir a escala de grises
gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

laplaciano = cv2.Laplacian(gris, cv2.CV_64F)
laplaciano = cv2.convertScaleAbs(laplaciano)


# Suavizado Gaussiano
gauss = cv2.GaussianBlur(gris, (3,3), 1)

# Aplicar Laplaciano al suavizado
log = cv2.Laplacian(gauss, cv2.CV_64F)
log = cv2.convertScaleAbs(log)


# Guardar Laplaciano
cv2.imwrite("Laplaciano.jpg", laplaciano)

# Guardar LoG
cv2.imwrite("Laplaciano_del_Gaussiano.jpg", log)

plt.figure(figsize=(12,18))

# Imagen original
plt.subplot(3,1,1)
plt.imshow(gris, cmap='gray')
plt.title("Imagen Original", fontsize=18)
plt.axis("off")

# Laplaciano
plt.subplot(3,1,2)
plt.imshow(laplaciano, cmap='gray')
plt.title("Detector Laplaciano", fontsize=18)
plt.axis("off")

# Laplaciano del Gaussiano
plt.subplot(3,1,3)
plt.imshow(log, cmap='gray')
plt.title("Laplaciano del Gaussiano (LoG)", fontsize=18)
plt.axis("off")

plt.tight_layout()
plt.show()
```
